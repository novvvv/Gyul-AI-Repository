```python
import pandas as pd
import matplotlib.pyplot as plt

# Font
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = [
    "Apple SD Gothic Neo",
    "AppleGothic",
    "Malgun Gothic",
    "NanumGothic",
    "NanumBarunGothic",
    "Noto Sans CJK KR",
    "DejaVu Sans",
]
plt.rcParams["axes.unicode_minus"] = False

CSV_PATH = "./5차년도_2차.csv"
df = pd.read_csv(CSV_PATH, encoding="cp949")

print("전체 행 수:", len(df))
print("컬럼:", list(df.columns))

print("\nSituation distribution")
display(df["상황"].value_counts(dropna=False).to_frame("count"))

print("\nSituation ratio (%)")
display((df["상황"].value_counts(normalize=True, dropna=False) * 100).round(2).to_frame("pct"))

df["상황"].value_counts(dropna=False).plot(
    kind="bar", figsize=(12, 4), title="Situation distribution", color="#11120D"
)
plt.xlabel("Situation")
plt.ylabel("Count", rotation=0, loc="top")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
